```python
#Notebook 1 — Python

# Get student details
name = input("Enter student name: ")
python_marks = int(input("Enter Python marks: "))
sql_marks = int(input("Enter SQL marks: "))

# Store marks in a list
marks = [python_marks, sql_marks]
subjects = ["Python", "SQL"]

# Calculate total and average
total_marks = sum(marks)
average_marks = total_marks / len(marks)

# Decide the grade
if average_marks >= 80:
    grade = "A"
elif average_marks >= 60:
    grade = "B"
elif average_marks >= 40:
    grade = "C"
else:
    grade = "F"

# Check data types using isinstance()
print("\nData Type Validation")
print("--------------------")
print("Name is string:", isinstance(name, str))
print("Marks are integers:", all(isinstance(mark, int) for mark in marks))

# Check marks using any() and all()
print("Any subject below 40:", any(mark < 40 for mark in marks))
print("All marks are passing:", all(mark >= 40 for mark in marks))

# Use map() to convert marks to strings
marks_text = list(map(str, marks))

# Use zip() to combine subjects and marks
subject_marks = list(zip(subjects, marks))

# Use enumerate() to display subjects with index
print("\nSubject Details")
print("----------------")
for index, (subject, mark) in enumerate(subject_marks, start=1):
    print(index, subjects[index - 1], ":", mark)

# Use sorted() to arrange marks in ascending order
print("\nSorted Marks:", sorted(marks))

# Use filter() to get passing marks
passing_marks = list(filter(lambda mark: mark >= 40, marks))
print("Passing Marks:", passing_marks)

# Use id() to display object identity
print("\nObject Identity")
print("----------------")
print("Name ID:", id(name))
print("Marks ID:", id(marks))

# Display the result
print("\nStudent Result")
print("--------------")
print("Name:", name)
print("Python Marks:", python_marks)
print("SQL Marks:", sql_marks)
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
print("Grade:", grade)
print("Marks as strings:", marks_text)
```

```python
#NoteBook2

from abc import ABC, abstractmethod

# Composition Class


class Address:

    def __init__(self, city, state):
        self.city = city
        self.state = state

    def display_address(self):
        print("City:", self.city)
        print("State:", self.state)


# ============================================
# Parent Class
# ============================================

class Person:

    # Class variable
    organization = "ABC Engineering College"

    def __init__(self, name, age):

        # Instance variables
        self.name = name
        self.age = age

    # Instance method
    def display_person(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Organization:", Person.organization)


# ============================================
# Abstract Class
# ============================================

class Report(ABC):

    @abstractmethod
    def generate_report(self):
        pass


# ============================================
# Student Class
# Inheritance + Encapsulation + Composition
# ============================================

class Student(Person, Report):

    def __init__(
        self,
        name,
        age,
        python_marks,
        sql_marks,
        city,
        state
    ):

        # Calling parent constructor
        super().__init__(name, age)

        # Instance variables
        self.python_marks = python_marks
        self.sql_marks = sql_marks

        # Private variable
        self.__average = 0

        # Composition
        self.address = Address(city, state)

    # Method
    def calculate_total(self):

        return self.python_marks + self.sql_marks

    # Method
    def calculate_average(self):

        total = self.calculate_total()

        self.__average = total / 2

        return self.__average

    # Encapsulation - getter
    def get_average(self):

        return self.__average

    # Polymorphic method / overriding abstract method
    def generate_report(self):

        print("\nGenerating Student Report...")

    # Display method
    def display_student(self):

        print("\nStudent Details")
        print("--------------------")

        self.display_person()

        print("Python Marks:", self.python_marks)
        print("SQL Marks:", self.sql_marks)

        total = self.calculate_total()
        average = self.calculate_average()

        print("Total Marks:", total)
        print("Average Marks:", average)

        self.address.display_address()

        self.generate_report()


# Create Object

student = Student(
    "Balaji",
    22,
    85,
    78,
    "Madurai",
    "Tamil Nadu"
)


# ============================================
# Display Student
# ============================================

student.display_student()
```

```python
#NoteBook3

# STUDENT RESULT MANAGEMENT
# EXCEPTION HANDLING DEMONSTRATION

# Custom Exception


class InvalidMarksError(Exception):
    """Raised when student marks are outside 0-100."""
    pass



# Function to Validate Marks


def validate_marks(marks, subject):

    if marks < 0 or marks > 100:
        raise InvalidMarksError(
            f"{subject} marks must be between 0 and 100."
        )



# Main Program

try:

    # Get student details
    name = input("Enter student name: ")

    python_marks = int(
        input("Enter Python marks: ")
    )

    sql_marks = int(
        input("Enter SQL marks: ")
    )

    # Validate marks
    validate_marks(
        python_marks,
        "Python"
    )

    validate_marks(
        sql_marks,
        "SQL"
    )

    # Assert valid input
    assert python_marks >= 0
    assert sql_marks >= 0

    # Calculate result
    total_marks = python_marks + sql_marks
    average_marks = total_marks / 2

    # Decide grade
    if average_marks >= 80:
        grade = "A"

    elif average_marks >= 60:
        grade = "B"

    elif average_marks >= 40:
        grade = "C"

    else:
        grade = "F"


except ValueError:

    print(
        "Error: Please enter numeric values for marks."
    )


except InvalidMarksError as error:

    print("Marks Validation Error:", error)


except AssertionError:

    print("Error: Marks validation failed.")


else:

    print("\nStudent Result")
    print("--------------------")
    print("Name:", name)
    print("Python Marks:", python_marks)
    print("SQL Marks:", sql_marks)
    print("Total Marks:", total_marks)
    print("Average Marks:", average_marks)
    print("Grade:", grade)


finally:

    print("\nProgram execution completed.")
```

```python
#note book 4 
#using logging

import logging
from logging.handlers import RotatingFileHandler

# Create logger
logger = logging.getLogger("student_application")
logger.setLevel(logging.DEBUG)

# Create console handler
console_handler = logging.StreamHandler()

# Create rotating file handler
file_handler = RotatingFileHandler(
    "student_application.log",
    maxBytes=5000,
    backupCount=3
)

# Create formatter
formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

# Apply formatter to handlers
console_handler.setFormatter(formatter)
file_handler.setFormatter(formatter)

# Add handlers to logger
logger.addHandler(console_handler)
logger.addHandler(file_handler)

# Program starts
logger.info("Student application started")

student_name = "Balaji"
python_marks = 85
sql_marks = 80

logger.debug(f"Student name: {student_name}")
logger.debug(f"Python marks: {python_marks}")
logger.debug(f"SQL marks: {sql_marks}")

try:
    total_marks = python_marks + sql_marks

    logger.info("Total marks calculated successfully")

    print("----- Student Report -----")
    print("Student Name:", student_name)
    print("Python Marks:", python_marks)
    print("SQL Marks:", sql_marks)
    print("Total Marks:", total_marks)

except Exception:
    logger.exception("Error while creating student report")

logger.info("Student application completed")
```

```python
import mysql.connector

connection = mysql.connector.connect(
    host="127.0.0.1",
    port=3306,
    user="root",
    password="balaji@123"
)

cursor = connection.cursor()

# Create database
cursor.execute("CREATE DATABASE IF NOT EXISTS student_db")
cursor.execute("USE student_db")

# Create students table
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(50) NOT NULL,
    age INT CHECK (age >= 18),
    city VARCHAR(50) DEFAULT 'Chennai'
)
""")

# Create courses table
cursor.execute("""
CREATE TABLE IF NOT EXISTS courses (
    course_id INT PRIMARY KEY,
    course_name VARCHAR(50) UNIQUE NOT NULL
)
""")

# Create marks table
cursor.execute("""
CREATE TABLE IF NOT EXISTS marks (
    mark_id INT PRIMARY KEY,
    student_id INT,
    course_id INT,
    marks INT CHECK (marks BETWEEN 0 AND 100),
    FOREIGN KEY (student_id) REFERENCES students(student_id),
    FOREIGN KEY (course_id) REFERENCES courses(course_id)
)
""")

print("Database and tables created successfully.")
```

```python
# Display students from Chennai, sorted by age
cursor.execute("""
SELECT student_id, student_name, age, city
FROM students
WHERE city = 'Chennai'
ORDER BY age DESC
LIMIT 5
""")

for row in cursor.fetchall():
    print(row)
```

```python
# Find courses with an average mark of at least 75
cursor.execute("""
SELECT
    courses.course_name,
    COUNT(marks.mark_id) AS total_records,
    AVG(marks.marks) AS average_marks
FROM courses
INNER JOIN marks
    ON courses.course_id = marks.course_id
GROUP BY courses.course_id, courses.course_name
HAVING AVG(marks.marks) >= 75
ORDER BY average_marks DESC
""")

for row in cursor.fetchall():
    print(row)
```

```python
# Display unique student cities
cursor.execute("""
SELECT DISTINCT city
FROM students
ORDER BY city
""")

for row in cursor.fetchall():
    print(row)
```

```python
# Display students, courses and marks
cursor.execute("""
SELECT
    students.student_name,
    courses.course_name,
    marks.marks
FROM students
INNER JOIN marks
    ON students.student_id = marks.student_id
INNER JOIN courses
    ON marks.course_id = courses.course_id
ORDER BY marks.marks DESC
""")

for row in cursor.fetchall():
    print(row)
```

```python
# Display all marks records and their student details
cursor.execute("""
SELECT
    students.student_name,
    marks.course_id,
    marks.marks
FROM students
RIGHT JOIN marks
    ON students.student_id = marks.student_id
""")

for row in cursor.fetchall():
    print(row)
```

```python
# Combine matching and non-matching rows from both sides
cursor.execute("""
SELECT
    students.student_name,
    marks.mark_id,
    marks.marks
FROM students
LEFT JOIN marks
    ON students.student_id = marks.student_id

UNION

SELECT
    students.student_name,
    marks.mark_id,
    marks.marks
FROM students
RIGHT JOIN marks
    ON students.student_id = marks.student_id
""")

for row in cursor.fetchall():
    print(row)
```

```python
# Combine every student with every course
cursor.execute("""
SELECT
    students.student_name,
    courses.course_name
FROM students
CROSS JOIN courses
ORDER BY students.student_name
""")

for row in cursor.fetchall():
    print(row)
```

```python
# Assign grades and find marks above the overall average
cursor.execute("""
SELECT
    students.student_name,
    courses.course_name,
    marks.marks,
    CASE
        WHEN marks.marks >= 80 THEN 'A'
        WHEN marks.marks >= 60 THEN 'B'
        WHEN marks.marks >= 40 THEN 'C'
        ELSE 'F'
    END AS grade
FROM students
INNER JOIN marks
    ON students.student_id = marks.student_id
INNER JOIN courses
    ON marks.course_id = courses.course_id
WHERE marks.marks > (
    SELECT AVG(marks)
    FROM marks
)
ORDER BY marks.marks DESC
""")

for row in cursor.fetchall():
    print(row)
```

```python
# Calculate student average marks and rank students
cursor.execute("""
WITH student_summary AS (
    SELECT
        students.student_id,
        students.student_name,
        AVG(marks.marks) AS average_marks
    FROM students
    INNER JOIN marks
        ON students.student_id = marks.student_id
    GROUP BY students.student_id, students.student_name
)
SELECT
    student_id,
    student_name,
    ROUND(average_marks, 2) AS average_marks,
    ROW_NUMBER() OVER (ORDER BY average_marks DESC) AS row_num,
    RANK() OVER (ORDER BY average_marks DESC) AS student_rank,
    DENSE_RANK() OVER (ORDER BY average_marks DESC) AS dense_rank_value,
    LAG(ROUND(average_marks, 2)) OVER (
        ORDER BY average_marks DESC
    ) AS previous_average,
    LEAD(ROUND(average_marks, 2)) OVER (
        ORDER BY average_marks DESC
    ) AS next_average
FROM student_summary
ORDER BY average_marks DESC
""")

for row in cursor.fetchall():
    print(row)
```

```python
# Create a reusable view
cursor.execute("""
CREATE OR REPLACE VIEW student_marks_view AS
SELECT
    students.student_name,
    courses.course_name,
    marks.marks
FROM students
INNER JOIN marks
    ON students.student_id = marks.student_id
INNER JOIN courses
    ON marks.course_id = courses.course_id
""")

# Retrieve data from the view
cursor.execute("SELECT * FROM student_marks_view")

for row in cursor.fetchall():
    print(row)
```
