```python
#Notebook 1 — Python

# Get student details
name = input("Enter student name: ")
python_marks = int(input("Enter Python marks: "))
sql_marks = int(input("Enter SQL marks: "))

# Store marks in a list
marks = [python_marks, sql_marks]
subjects = ["Python", "SQL"]

# Calculate total and average
total_marks = sum(marks)
average_marks = total_marks / len(marks)

# Decide the grade
if average_marks >= 80:
    grade = "A"
elif average_marks >= 60:
    grade = "B"
elif average_marks >= 40:
    grade = "C"
else:
    grade = "F"

# Check data types using isinstance()
print("\nData Type Validation")
print("--------------------")
print("Name is string:", isinstance(name, str))
print("Marks are integers:", all(isinstance(mark, int) for mark in marks))

# Check marks using any() and all()
print("Any subject below 40:", any(mark < 40 for mark in marks))
print("All marks are passing:", all(mark >= 40 for mark in marks))

# Use map() to convert marks to strings
marks_text = list(map(str, marks))

# Use zip() to combine subjects and marks
subject_marks = list(zip(subjects, marks))

# Use enumerate() to display subjects with index
print("\nSubject Details")
print("----------------")
for index, (subject, mark) in enumerate(subject_marks, start=1):
    print(index, subjects[index - 1], ":", mark)

# Use sorted() to arrange marks in ascending order
print("\nSorted Marks:", sorted(marks))

# Use filter() to get passing marks
passing_marks = list(filter(lambda mark: mark >= 40, marks))
print("Passing Marks:", passing_marks)

# Use id() to display object identity
print("\nObject Identity")
print("----------------")
print("Name ID:", id(name))
print("Marks ID:", id(marks))

# Display the result
print("\nStudent Result")
print("--------------")
print("Name:", name)
print("Python Marks:", python_marks)
print("SQL Marks:", sql_marks)
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
print("Grade:", grade)
print("Marks as strings:", marks_text)
```

```python
#NoteBook2

from abc import ABC, abstractmethod

# Composition Class


class Address:

    def __init__(self, city, state):
        self.city = city
        self.state = state

    def display_address(self):
        print("City:", self.city)
        print("State:", self.state)


# ============================================
# Parent Class
# ============================================

class Person:

    # Class variable
    organization = "ABC Engineering College"

    def __init__(self, name, age):

        # Instance variables
        self.name = name
        self.age = age

    # Instance method
    def display_person(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Organization:", Person.organization)


# ============================================
# Abstract Class
# ============================================

class Report(ABC):

    @abstractmethod
    def generate_report(self):
        pass


# ============================================
# Student Class
# Inheritance + Encapsulation + Composition
# ============================================

class Student(Person, Report):

    def __init__(
        self,
        name,
        age,
        python_marks,
        sql_marks,
        city,
        state
    ):

        # Calling parent constructor
        super().__init__(name, age)

        # Instance variables
        self.python_marks = python_marks
        self.sql_marks = sql_marks

        # Private variable
        self.__average = 0

        # Composition
        self.address = Address(city, state)

    # Method
    def calculate_total(self):

        return self.python_marks + self.sql_marks

    # Method
    def calculate_average(self):

        total = self.calculate_total()

        self.__average = total / 2

        return self.__average

    # Encapsulation - getter
    def get_average(self):

        return self.__average

    # Polymorphic method / overriding abstract method
    def generate_report(self):

        print("\nGenerating Student Report...")

    # Display method
    def display_student(self):

        print("\nStudent Details")
        print("--------------------")

        self.display_person()

        print("Python Marks:", self.python_marks)
        print("SQL Marks:", self.sql_marks)

        total = self.calculate_total()
        average = self.calculate_average()

        print("Total Marks:", total)
        print("Average Marks:", average)

        self.address.display_address()

        self.generate_report()


# Create Object

student = Student(
    "Balaji",
    22,
    85,
    78,
    "Madurai",
    "Tamil Nadu"
)


# ============================================
# Display Student
# ============================================

student.display_student()
```

```python
#NoteBook3

# STUDENT RESULT MANAGEMENT
# EXCEPTION HANDLING DEMONSTRATION

# Custom Exception


class InvalidMarksError(Exception):
    """Raised when student marks are outside 0-100."""
    pass



# Function to Validate Marks


def validate_marks(marks, subject):

    if marks < 0 or marks > 100:
        raise InvalidMarksError(
            f"{subject} marks must be between 0 and 100."
        )



# Main Program

try:

    # Get student details
    name = input("Enter student name: ")

    python_marks = int(
        input("Enter Python marks: ")
    )

    sql_marks = int(
        input("Enter SQL marks: ")
    )

    # Validate marks
    validate_marks(
        python_marks,
        "Python"
    )

    validate_marks(
        sql_marks,
        "SQL"
    )

    # Assert valid input
    assert python_marks >= 0
    assert sql_marks >= 0

    # Calculate result
    total_marks = python_marks + sql_marks
    average_marks = total_marks / 2

    # Decide grade
    if average_marks >= 80:
        grade = "A"

    elif average_marks >= 60:
        grade = "B"

    elif average_marks >= 40:
        grade = "C"

    else:
        grade = "F"


except ValueError:

    print(
        "Error: Please enter numeric values for marks."
    )


except InvalidMarksError as error:

    print("Marks Validation Error:", error)


except AssertionError:

    print("Error: Marks validation failed.")


else:

    print("\nStudent Result")
    print("--------------------")
    print("Name:", name)
    print("Python Marks:", python_marks)
    print("SQL Marks:", sql_marks)
    print("Total Marks:", total_marks)
    print("Average Marks:", average_marks)
    print("Grade:", grade)


finally:

    print("\nProgram execution completed.")
```

```python
#note book 4 
#using logging

import logging
from logging.handlers import RotatingFileHandler

# Create logger
logger = logging.getLogger("student_application")
logger.setLevel(logging.DEBUG)

# Create console handler
console_handler = logging.StreamHandler()

# Create rotating file handler
file_handler = RotatingFileHandler(
    "student_application.log",
    maxBytes=5000,
    backupCount=3
)

# Create formatter
formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

# Apply formatter to handlers
console_handler.setFormatter(formatter)
file_handler.setFormatter(formatter)

# Add handlers to logger
logger.addHandler(console_handler)
logger.addHandler(file_handler)

# Program starts
logger.info("Student application started")

student_name = "Balaji"
python_marks = 85
sql_marks = 80

logger.debug(f"Student name: {student_name}")
logger.debug(f"Python marks: {python_marks}")
logger.debug(f"SQL marks: {sql_marks}")

try:
    total_marks = python_marks + sql_marks

    logger.info("Total marks calculated successfully")

    print("----- Student Report -----")
    print("Student Name:", student_name)
    print("Python Marks:", python_marks)
    print("SQL Marks:", sql_marks)
    print("Total Marks:", total_marks)

except Exception:
    logger.exception("Error while creating student report")

logger.info("Student application completed")
```
